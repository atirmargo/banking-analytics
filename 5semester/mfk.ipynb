```python
import sklearn
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

random_seed = 4238
np.random.seed(random_seed)

X, y = load_breast_cancer(return_X_y=True)
X_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, shuffle=True, random_state=42
)

clf = KNeighborsClassifier(n_neighbors=8, p=1)
clf.fit(X_train, y_train)

predictions = clf.predict(x_test)
acc = accuracy_score(y_test, predictions)
acc = np.floor(acc * 100) / 100  # округление вниз до сотых

print(acc)
```

```python
import sklearn
import numpy as np
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

random_seed = 4238
np.random.seed(random_seed)
n_splits = 3

from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)

metrics = ['manhattan', 'euclidean', 'cosine']
best_score = 0
best_metric = None

for metric in metrics:
    cv_scores = []
    for k in range(1, 51):
        clf = KNeighborsClassifier(n_neighbors=k, metric=metric)
        scores = cross_val_score(clf, X, y, cv=n_splits, scoring='accuracy')
        cv_scores.append(scores.mean())
    max_score = max(cv_scores)
    print(f"Metric: {metric}, best K = {np.argmax(cv_scores)+1}, best score = {max_score:.4f}")
    if max_score > best_score:
        best_score = max_score
        best_metric = metric

print(f"Best metric: {best_metric} with score = {best_score:.4f}")
```

```python
N, C, R = map(int, input().split())
forget = list(map(int, input().split()))
spare = list(map(int, input().split()))

forget_set = set(forget)
spare_set = set(spare)

# 0 - нет карты, 1 - есть карта 
has_card = [1] * (N + 1)  # изначально у всех основная
for i in forget:
    has_card[i] = 0  # кто забыл, у тех нет основной

# Запасные карты
# Сначала раздаем тем, кто забыл и имеет запасную 
used_spare = [False] * (N + 1)

for person in spare:
    if person in forget_set:
        has_card[person] = 1
        used_spare[person] = True

# Остались те, у кого есть запасная, но они не забыли
# Они могут отдать соседям, которые забыли
changed = True
while changed:
    changed = False
    for person in spare:
        if used_spare[person]:
            continue  # карта уже использована
        # Соседи
        neighbors = []
        if person > 1:
            neighbors.append(person - 1)
        if person < N:
            neighbors.append(person + 1)
        for nb in neighbors:
            if has_card[nb] == 0:  # сосед забыл и еще без карты
                has_card[nb] = 1
                used_spare[person] = True
                changed = True
                break  

result = sum(has_card[1:])  
print(result)
```
